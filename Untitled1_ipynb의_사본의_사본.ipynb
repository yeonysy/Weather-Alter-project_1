```python
import cv2
import numpy as np
import torch
import segmentation_models_pytorch as smp
import easyocr
import ezdxf
import matplotlib.pyplot as plt

# 1. 전처리 함수
def preprocess_image(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    denoised = cv2.bilateralFilter(gray, 9, 75, 75)
    binary = cv2.adaptiveThreshold(
        denoised, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY_INV, 15, 2
    )
    return binary

# 2. AI 모델 설정
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = smp.Unet(encoder_name="resnet34", encoder_weights="imagenet", in_channels=3, classes=4).to(device)
model.eval()

def predict_wall_mask(model, image_bgr):
    img_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
    resized = cv2.resize(img_rgb, (512, 512))
    tensor_img = torch.tensor(resized).permute(2, 0, 1).unsqueeze(0).float() / 255.0
    with torch.no_grad():
        output = model(tensor_img.to(device))
        preds = torch.argmax(output, dim=1).squeeze(0).cpu().numpy()
    return cv2.resize(preds.astype(np.uint8), (image_bgr.shape[1], image_bgr.shape[0]), interpolation=cv2.INTER_NEAREST)

# 3. OCR 및 DXF 내보내기
reader = easyocr.Reader(['ko', 'en'], gpu=torch.cuda.is_available())

def export_to_dxf(binary_mask, text_data, output_path="result.dxf"):
    doc = ezdxf.new(dxfversion="AC1027")
    msp = doc.modelspace()
    doc.layers.add("WALLS", color=1)
    doc.layers.add("TEXT_DIM", color=3)

    contours, _ = cv2.findContours(binary_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    for cnt in contours:
        approx = cv2.approxPolyDP(cnt, 0.005 * cv2.arcLength(cnt, True), True)
        if len(approx) >= 2:
            pts = [(float(p[0][0]), float(-p[0][1])) for p in approx]
            msp.add_lwpolyline(pts, close=True, dxfattribs={"layer": "WALLS"})

    for item in text_data:
        x, y = float(item[0][0][0]), float(-item[0][0][1])
        msp.add_text(item[1], dxfattribs={"layer": "TEXT_DIM", "height": 15}).set_placement((x, y))

    doc.saveas(output_path)

# 4. 테스트용 도면 생성 및 파이프라인 실행
test_img = np.ones((800, 800, 3), dtype=np.uint8) * 255
cv2.rectangle(test_img, (100, 100), (700, 700), (0, 0, 0), 12)
cv2.line(test_img, (400, 100), (400, 700), (0, 0, 0), 8)
cv2.putText(test_img, "ROOM A", (180, 200), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 0), 2)
cv2.putText(test_img, "4500", (220, 130), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 0, 0), 2)

binary_img = preprocess_image(test_img)
predicted_mask = predict_wall_mask(model, test_img)
ocr_results = reader.readtext(test_img)

export_to_dxf(binary_img, ocr_results, "result_floorplan.dxf")

# 결과 시각화
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(cv2.cvtColor(test_img, cv2.COLOR_BGR2RGB)); axes[0].set_title("1. Original")
axes[1].imshow(binary_img, cmap='gray'); axes[1].set_title("2. Preprocessed Binary")
axes[2].imshow(predicted_mask, cmap='jet'); axes[2].set_title("3. AI Mask")
plt.show()

print("🎉 실행 완료! 왼쪽 폴더 모양(파일) 탭에서 'result_floorplan.dxf'를 다운로드할 수 있습니다.")
```

```python
# 1. Gradio 및 필수 패키지 설치
!pip install -q gradio ezdxf easyocr opencv-python-headless torch segmentation-models-pytorch

import cv2
import numpy as np
import ezdxf
import easyocr
import torch
import gradio as gr

# OCR 및 디바이스 세팅
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
reader = easyocr.Reader(['ko', 'en'], gpu=torch.cuda.is_available())

def process_floorplan(input_image):
    """사용자가 업로드한 이미지를 받아 DXF 변환 후 결과 반환"""
    if input_image is None:
        return None, None, "이미지를 업로드해주세요."

    # 1. OpenCV 이미지 전처리
    gray = cv2.cvtColor(input_image, cv2.COLOR_RGB2GRAY)
    denoised = cv2.bilateralFilter(gray, 9, 75, 75)
    binary = cv2.adaptiveThreshold(
        denoised, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY_INV, 15, 2
    )

    # 2. OCR 텍스트 인식
    ocr_results = reader.readtext(input_image)

    # 3. DXF 파일 생성
    dxf_path = "output_converted.dxf"
    doc = ezdxf.new(dxfversion="AC1027")
    msp = doc.modelspace()
    doc.layers.add("WALLS", color=1)      # 벽체 (빨간색)
    doc.layers.add("TEXT_DIM", color=3)   # 텍스트 (초록색)

    # 외곽선(벽체) 추출 및 DXF 작성
    contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    for cnt in contours:
        approx = cv2.approxPolyDP(cnt, 0.005 * cv2.arcLength(cnt, True), True)
        if len(approx) >= 2:
            pts = [(float(p[0][0]), float(-p[0][1])) for p in approx]
            msp.add_lwpolyline(pts, close=True, dxfattribs={"layer": "WALLS"})

    # OCR 텍스트 DXF 추가
    for item in ocr_results:
        x, y = float(item[0][0][0]), float(-item[0][0][1])
        msp.add_text(item[1], dxfattribs={"layer": "TEXT_DIM", "height": 15}).set_placement((x, y))

    doc.saveas(dxf_path)

    # 미리보기용 이미지
    preview_img = cv2.cvtColor(binary, cv2.COLOR_GRAY2RGB)

    return preview_img, dxf_path, f"변환 완료! {len(contours)}개 구조 요소 및 {len(ocr_results)}개 텍스트 추출됨."

# Gradio 웹 데모 인터페이스 구성 (최신 Gradio 호환 파라미터 적용)
demo = gr.Interface(
    fn=process_floorplan,
    inputs=gr.Image(type="numpy", label="도면 이미지 업로드 (JPG/PNG)"),
    outputs=[
        gr.Image(label="AI 이진화 미리보기"),
        gr.File(label="변환된 CAD (DXF) 파일 다운로드"),
        gr.Textbox(label="처리 상태")
    ],
    title="🏢 PDF/이미지 도면 ➔ CAD(DXF) 변환 AI 서비스 MVP",
    description="도면 이미지 파일(PNG/JPG)을 업로드하면 AI가 벽체와 텍스트를 인식하여 수정 가능한 CAD(.dxf) 파일로 자동 변환해 줍니다."
)

# Colab 및 외부 접속용 링크 생성
demo.launch(share=True)
```
